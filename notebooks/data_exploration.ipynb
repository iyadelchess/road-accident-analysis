# Exploration des données

Cette première étape consiste à explorer et comprendre la structure des données avant de commencer leur nettoyage et leur analyse.

## Jeux de données utilisés

Les données sont réparties en quatre fichiers :

- `caract-2024.csv` : caractéristiques générales des accidents ;
- `lieux-2024.csv` : informations sur les lieux et les routes ;
- `vehicules-2024.csv` : informations sur les véhicules impliqués ;
- `usagers-2024.csv` : informations sur les usagers impliqués.

```python
import pandas as pd
```

```python
# Chargement des quatre jeux de données
caracteristiques = pd.read_csv(
    "../data/raw/caract-2024.csv",
    sep=";"
)

lieux = pd.read_csv(
    "../data/raw/lieux-2024.csv",
    sep=";"
)

vehicules = pd.read_csv(
    "../data/raw/vehicules-2024.csv",
    sep=";"
)

usagers = pd.read_csv(
    "../data/raw/usagers-2024.csv",
    sep=";"
)
```

```python
# Regroupement des jeux de données
datasets = {
    "Caractéristiques": caracteristiques,
    "Lieux": lieux,
    "Véhicules": vehicules,
    "Usagers": usagers
}

# Affichage des dimensions
for name, df in datasets.items():
    print(f"{name} : {df.shape[0]} lignes, {df.shape[1]} colonnes")
```

## Première observation

Les données sont réparties en quatre jeux de données de tailles différentes.

Le fichier des caractéristiques contient 54 402 lignes, tandis que les fichiers concernant les véhicules et les usagers contiennent davantage d'enregistrements.

Cela indique qu'un accident peut être associé à plusieurs véhicules et à plusieurs usagers.

```python
# Affichage des colonnes de chaque jeu de données
for name, df in datasets.items():
    print(f"\n{name}")
    print(df.columns.tolist())
```

## Structure des données

Les quatre jeux de données décrivent différents aspects des accidents de la route.

La colonne `Num_Acc` est présente dans les quatre fichiers et permet d'identifier un accident. Elle pourra donc être utilisée pour relier les différents jeux de données.

Cependant, les données ne présentent pas toutes le même niveau de granularité : un accident peut concerner plusieurs véhicules et plusieurs usagers. Les opérations de fusion devront donc être réalisées avec précaution.

```python
# Affichage du nombre de valeurs manquantes
for name, df in datasets.items():
    print(f"\n{name}")
    print(df.isnull().sum())
```

## Valeurs manquantes

Certaines variables contiennent des valeurs manquantes.

Par exemple, la colonne `adr` contient des valeurs manquantes dans le jeu de données des caractéristiques, tandis que `an_nais` est parfois manquante dans le jeu de données des usagers.

Certaines colonnes présentent également un nombre très important de valeurs manquantes. Leur pertinence devra être évaluée avant de les utiliser dans l'analyse.

## Conclusion

Cette première exploration a permis de comprendre la structure générale des quatre jeux de données.

Les données sont réparties selon différents niveaux d'information : accident, lieu, véhicule et usager.
